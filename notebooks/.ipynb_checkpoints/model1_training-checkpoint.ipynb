# SmartCHDR — Model 1: Growth and Weight Predictor

## Purpose
Train a Random Forest model to:
1. Classify child growth status: normal, underweight, overweight, stunted
2. Predict expected weight and height based on WHO standards

## Input Features
- age_months, gender, weight_kg, height_cm

## Output
- growth_status (classifier)
- predicted weight and height vs WHO median (regressor)

## Dataset
- Source: WHO Child Growth Standards (synthetic, 12,200 records)
- File: datasets/model1_growth/model1_growth_dataset.csv

```python
import subprocess
subprocess.run(["pip", "install", "openpyxl"], check=True)
print("✅ Done!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle

from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (classification_report, confusion_matrix, 
                              accuracy_score, mean_absolute_error, r2_score)

np.random.seed(42)
print("✅ Libraries imported!")
```

```python
# Load dataset
df = pd.read_csv("../datasets/model1_growth/model1_growth_dataset.csv")

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn info:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nGrowth status distribution:")
print(df['growth_status'].value_counts())
print("\nBasic statistics:")
print(df.describe())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Model 1 — Dataset Overview', fontsize=14, fontweight='bold')

# Chart 1: Growth status distribution
colors = ['#2ecc71','#e74c3c','#f39c12','#3498db']
df['growth_status'].value_counts().plot(kind='bar', ax=axes[0], 
                                         color=colors, edgecolor='black')
axes[0].set_title('Growth Status Distribution')
axes[0].set_xlabel('Status')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

# Chart 2: Weight distribution by status
for status, color in zip(['normal','stunted','underweight','overweight'], colors):
    subset = df[df['growth_status']==status]
    axes[1].scatter(subset['age_months'], subset['weight_kg'], 
                    c=color, label=status, alpha=0.3, s=8)
axes[1].set_title('Weight vs Age by Status')
axes[1].set_xlabel('Age (months)')
axes[1].set_ylabel('Weight (kg)')
axes[1].legend(markerscale=3)

# Chart 3: Height distribution by status
for status, color in zip(['normal','stunted','underweight','overweight'], colors):
    subset = df[df['growth_status']==status]
    axes[2].scatter(subset['age_months'], subset['height_cm'],
                    c=color, label=status, alpha=0.3, s=8)
axes[2].set_title('Height vs Age by Status')
axes[2].set_xlabel('Age (months)')
axes[2].set_ylabel('Height (cm)')
axes[2].legend(markerscale=3)

plt.tight_layout()
plt.savefig('../datasets/model1_data_overview.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Charts saved!")
```

```python
# Copy dataframe
df_model = df.copy()

# Encode gender: male=1, female=0
df_model['gender_encoded'] = (df_model['gender'] == 'male').astype(int)

# Encode growth status labels
le = LabelEncoder()
df_model['status_encoded'] = le.fit_transform(df_model['growth_status'])

print("Growth status encoding:")
for i, label in enumerate(le.classes_):
    print(f"  {i} = {label}")

# Define features and targets
# Features used to make predictions
FEATURES = ['age_months', 'gender_encoded', 'weight_kg', 'height_cm',
            'weight_zscore', 'height_zscore']

# Target 1: growth status classification
X = df_model[FEATURES]
y_class = df_model['status_encoded']

# Target 2: regression — predict weight and height WHO medians
y_weight = df_model['weight_median_who']
y_height = df_model['height_median_who']

# Split: 80% training, 20% testing
X_train, X_test, y_class_train, y_class_test = train_test_split(
    X, y_class, test_size=0.2, random_state=42, stratify=y_class)

_, _, y_weight_train, y_weight_test = train_test_split(
    X, y_weight, test_size=0.2, random_state=42)

_, _, y_height_train, y_height_test = train_test_split(
    X, y_height, test_size=0.2, random_state=42)

print(f"\nTraining set size: {len(X_train):,}")
print(f"Testing set size:  {len(X_test):,}")
print(f"\nFeatures used: {FEATURES}")
print("✅ Data prepared!")
```

```python
print("Training Random Forest Classifier...")
print("(This may take 30-60 seconds)")

# Create and train the classifier
clf = RandomForestClassifier(
    n_estimators=100,    # 100 decision trees
    max_depth=15,        # Maximum depth of each tree
    min_samples_split=5, # Minimum samples to split a node
    random_state=42,
    n_jobs=-1            # Use all CPU cores
)

clf.fit(X_train, y_class_train)

# Evaluate on test set
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_class_test, y_pred)

print(f"\n✅ Classifier trained!")
print(f"Accuracy: {accuracy*100:.2f}%")
print(f"\nDetailed Report:")
print(classification_report(y_class_test, y_pred, 
                             target_names=le.classes_))
```

```python
cm = confusion_matrix(y_class_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=le.classes_,
            yticklabels=le.classes_)
plt.title('Model 1 — Confusion Matrix\n(Growth Status Classifier)', 
          fontsize=13, fontweight='bold')
plt.ylabel('Actual Status')
plt.xlabel('Predicted Status')
plt.tight_layout()
plt.savefig('../datasets/model1_confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Confusion matrix saved!")
```

```python
print("Running 5-fold cross validation...")
cv_scores = cross_val_score(clf, X, y_class, cv=5, scoring='accuracy', n_jobs=-1)

print(f"\nCross Validation Results:")
print(f"Scores per fold: {[f'{s*100:.2f}%' for s in cv_scores]}")
print(f"Mean accuracy:   {cv_scores.mean()*100:.2f}%")
print(f"Std deviation:   {cv_scores.std()*100:.2f}%")

if cv_scores.mean() > 0.85:
    print("\n✅ Excellent! Model is performing very well.")
elif cv_scores.mean() > 0.75:
    print("\n✅ Good! Model is performing well.")
else:
    print("\n⚠️ Model may need improvement.")
```

```python
importance_df = pd.DataFrame({
    'feature': FEATURES,
    'importance': clf.feature_importances_
}).sort_values('importance', ascending=True)

plt.figure(figsize=(8, 5))
plt.barh(importance_df['feature'], importance_df['importance'], 
         color='#3498db', edgecolor='black')
plt.title('Model 1 — Feature Importance\n(Which features matter most?)',
          fontsize=13, fontweight='bold')
plt.xlabel('Importance Score')
plt.tight_layout()
plt.savefig('../datasets/model1_feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()
print("\nFeature importance ranking:")
for _, row in importance_df.iloc[::-1].iterrows():
    print(f"  {row['feature']:25s}: {row['importance']:.4f}")
```

```python
print("Training Weight Regressor...")
reg_weight = RandomForestRegressor(
    n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
reg_weight.fit(X_train, y_weight_train)

print("Training Height Regressor...")
reg_height = RandomForestRegressor(
    n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
reg_height.fit(X_train, y_height_train)

# Evaluate regressors
w_pred = reg_weight.predict(X_test)
h_pred = reg_height.predict(X_test)

w_mae = mean_absolute_error(y_weight_test, w_pred)
h_mae = mean_absolute_error(y_height_test, h_pred)
w_r2  = r2_score(y_weight_test, w_pred)
h_r2  = r2_score(y_height_test, h_pred)

print(f"\n✅ Regressors trained!")
print(f"\nWeight Regressor:")
print(f"  Mean Absolute Error: {w_mae:.3f} kg")
print(f"  R² Score:            {w_r2:.4f}")
print(f"\nHeight Regressor:")
print(f"  Mean Absolute Error: {h_mae:.3f} cm")
print(f"  R² Score:            {h_r2:.4f}")
```

```python
# Example: 12 month old boy, weight 8.5kg, height 72cm
sample = pd.DataFrame([{
    'age_months'    : 12,
    'gender_encoded': 1,        # male
    'weight_kg'     : 8.5,
    'height_cm'     : 72.0,
    'weight_zscore' : -0.8,
    'height_zscore' : -0.5
}])

# Predict
predicted_status_encoded = clf.predict(sample)[0]
predicted_status = le.inverse_transform([predicted_status_encoded])[0]
predicted_weight = reg_weight.predict(sample)[0]
predicted_height = reg_height.predict(sample)[0]
probabilities    = clf.predict_proba(sample)[0]

print("=" * 45)
print("SAMPLE PREDICTION — 12 month old boy")
print("=" * 45)
print(f"Input:")
print(f"  Age:           12 months")
print(f"  Gender:        Male")
print(f"  Weight:        8.5 kg")
print(f"  Height:        72.0 cm")
print(f"\nPredictions:")
print(f"  Growth Status: {predicted_status.upper()}")
print(f"  WHO Expected Weight: {predicted_weight:.2f} kg")
print(f"  WHO Expected Height: {predicted_height:.2f} cm")
print(f"\nConfidence:")
for label, prob in zip(le.classes_, probabilities):
    print(f"  {label:12s}: {prob*100:.1f}%")
```

```python
# Create ml_models folder if it doesn't exist
os.makedirs('../ml_models', exist_ok=True)

# Save classifier
with open('../ml_models/model1_growth_classifier.pkl', 'wb') as f:
    pickle.dump(clf, f)

# Save regressors
with open('../ml_models/model1_weight_regressor.pkl', 'wb') as f:
    pickle.dump(reg_weight, f)

with open('../ml_models/model1_height_regressor.pkl', 'wb') as f:
    pickle.dump(reg_height, f)

# Save label encoder (needed to decode predictions back to text)
with open('../ml_models/model1_label_encoder.pkl', 'wb') as f:
    pickle.dump(le, f)

# Save feature list (important for Flask API to use same features)
with open('../ml_models/model1_features.pkl', 'wb') as f:
    pickle.dump(FEATURES, f)

print("✅ All Model 1 files saved to ml_models/:")
print("  - model1_growth_classifier.pkl")
print("  - model1_weight_regressor.pkl")
print("  - model1_height_regressor.pkl")
print("  - model1_label_encoder.pkl")
print("  - model1_features.pkl")

# Confirm file sizes
for fname in ['model1_growth_classifier.pkl', 'model1_weight_regressor.pkl',
              'model1_height_regressor.pkl', 'model1_label_encoder.pkl',
              'model1_features.pkl']:
    size = os.path.getsize(f'../ml_models/{fname}')
    print(f"  {fname}: {size/1024:.1f} KB")
```

## Model 1 Training Complete ✅

### Models Saved
| File | Purpose |
|---|---|
| model1_growth_classifier.pkl | Classifies growth status |
| model1_weight_regressor.pkl | Predicts expected WHO weight |
| model1_height_regressor.pkl | Predicts expected WHO height |
| model1_label_encoder.pkl | Decodes numeric labels to text |
| model1_features.pkl | Feature list for Flask API |

### How Flask Will Use These
1. Load all .pkl files when Flask starts
2. Accept age, gender, weight, height from frontend
3. Run classifier → return growth status
4. Run regressors → return expected weight and height
5. Return all results as JSON to React frontend
